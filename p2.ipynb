```python
import numpy as np
import random

cities = {
    "A": (0, 0),
    "B": (2, 5),
    "C": (5, 2),
    "D": (6, 6),
    "E": (8, 3),
    "F": (1, 7),
    "G": (3, 9),
    "H": (9, 8),
    "I": (7, 10),
    "J": (10, 5)
}

num_ants = 10
alpha = 1.0
beta = 2.0
rho = 0.5
tau0 = 1.0
max_iterations = 100
Q = 100

random.seed(42)
np.random.seed(42)

city_names = list(cities.keys())
n = len(city_names)

distance = np.zeros((n, n))

for i in range(n):
    for j in range(n):
        if i != j:
            x1, y1 = cities[city_names[i]]
            x2, y2 = cities[city_names[j]]
            distance[i][j] = np.sqrt(
                (x2 - x1) ** 2 + (y2 - y1) ** 2
            )

pheromone = np.full((n, n), tau0)
np.fill_diagonal(pheromone, 0)

heuristic = np.zeros((n, n))

for i in range(n):
    for j in range(n):
        if i != j:
            heuristic[i][j] = 1 / distance[i][j]

best_route = None
best_distance = float("inf")

for iteration in range(max_iterations):
    all_routes = []
    all_distances = []

    for ant in range(num_ants):
        start = random.randrange(n)
        route = [start]
        unvisited = set(range(n))
        unvisited.remove(start)

        while unvisited:
            current = route[-1]
            candidates = list(unvisited)

            weights = [
                (pheromone[current][city] ** alpha)
                * (heuristic[current][city] ** beta)
                for city in candidates
            ]

            next_city = random.choices(
                candidates, weights=weights, k=1
            )[0]

            route.append(next_city)
            unvisited.remove(next_city)

        route.append(start)

        total_distance = sum(
            distance[route[i]][route[i + 1]]
            for i in range(len(route) - 1)
        )

        all_routes.append(route)
        all_distances.append(total_distance)

        if total_distance < best_distance:
            best_distance = total_distance
            best_route = route.copy()

    pheromone *= (1 - rho)

    for route, route_distance in zip(all_routes, all_distances):
        deposit = Q / route_distance

        for i in range(len(route) - 1):
            a = route[i]
            b = route[i + 1]

            pheromone[a][b] += deposit
            pheromone[b][a] += deposit

best_city_route = [city_names[i] for i in best_route]

print("Best Tourist Tour:", " -> ".join(best_city_route))
print("Minimum Distance:", round(best_distance, 2))
print("Number of Cities:", n)
print("Number of Ants:", num_ants)
print("Maximum Iterations:", max_iterations)
```
